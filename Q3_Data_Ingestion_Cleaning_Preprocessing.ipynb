```python
import pandas as pd

df = pd.read_csv("retail_sales_raw_10000.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
# Check dataset information before cleaning

print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Create a messy copy for cleaning practice

messy_df = df.copy()

# Add missing values
messy_df.loc[10, "Sales"] = np.nan
messy_df.loc[20, "Profit"] = np.nan
messy_df.loc[30, "City"] = np.nan

# Add duplicate records
messy_df = pd.concat(
    [messy_df, messy_df.iloc[[0, 1, 2]]],
    ignore_index=True
)

# Create inconsistent data types
messy_df.loc[40, "Quantity"] = np.nan
messy_df.loc[50, "Discount"] = np.nan

print("Messy dataset created!")
print("Rows:", messy_df.shape[0])
print("Columns:", messy_df.shape[1])

messy_df.head()
```

```python
print("Missing Values Before Cleaning:")
print(messy_df.isnull().sum())

print("\nDuplicate Rows Before Cleaning:")
print(messy_df.duplicated().sum())
```

```python
# Handle missing values

# Numeric columns → fill with median
messy_df["Sales"] = messy_df["Sales"].fillna(messy_df["Sales"].median())
messy_df["Profit"] = messy_df["Profit"].fillna(messy_df["Profit"].median())
messy_df["Quantity"] = messy_df["Quantity"].fillna(messy_df["Quantity"].median())
messy_df["Discount"] = messy_df["Discount"].fillna(0)

# Text column → fill with mode
messy_df["City"] = messy_df["City"].fillna(messy_df["City"].mode()[0])

print("Missing values handled successfully!")
print("\nRemaining missing values:")
print(messy_df.isnull().sum())
```

```python
# Remove duplicate records

before_duplicates = messy_df.shape[0]

messy_df = messy_df.drop_duplicates()

after_duplicates = messy_df.shape[0]

print("Duplicate records removed:", before_duplicates - after_duplicates)
print("Rows after removing duplicates:", after_duplicates)
print("Remaining duplicates:", messy_df.duplicated().sum())
```

```python
# Convert date columns to datetime

messy_df["Order Date"] = pd.to_datetime(
    messy_df["Order Date"],
    errors="coerce"
)

messy_df["Ship Date"] = pd.to_datetime(
    messy_df["Ship Date"],
    errors="coerce"
)

# Convert numeric columns to proper numeric type

messy_df["Sales"] = pd.to_numeric(
    messy_df["Sales"],
    errors="coerce"
)

messy_df["Quantity"] = pd.to_numeric(
    messy_df["Quantity"],
    errors="coerce"
)

messy_df["Discount"] = pd.to_numeric(
    messy_df["Discount"],
    errors="coerce"
)

messy_df["Profit"] = pd.to_numeric(
    messy_df["Profit"],
    errors="coerce"
)

print("Data types corrected successfully!\n")
print(messy_df.dtypes)
```

```python
# Detect outliers using IQR method

def check_outliers(column):
    Q1 = messy_df[column].quantile(0.25)
    Q3 = messy_df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = messy_df[
        (messy_df[column] < lower_limit) |
        (messy_df[column] > upper_limit)
    ]

    print(column, "outliers:", len(outliers))
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print()

check_outliers("Sales")
check_outliers("Profit")
```

```python
# Handle outliers using IQR capping

def cap_outliers(column):
    Q1 = messy_df[column].quantile(0.25)
    Q3 = messy_df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    messy_df[column] = messy_df[column].clip(
        lower_limit,
        upper_limit
    )

# Apply IQR capping
cap_outliers("Sales")
cap_outliers("Profit")

print("Outliers handled successfully!")
```

```python
# Feature Engineering

# Extract Year and Month from Order Date
messy_df["Order Year"] = messy_df["Order Date"].dt.year
messy_df["Order Month"] = messy_df["Order Date"].dt.month

# Calculate Profit Margin
messy_df["Profit Margin (%)"] = np.where(
    messy_df["Sales"] != 0,
    (messy_df["Profit"] / messy_df["Sales"]) * 100,
    0
)

print("Feature engineering completed!")

print("\nNew Columns:")
print(messy_df[["Order Year", "Order Month", "Profit Margin (%)"]].head())
```

```python
# Final validation after cleaning

print("===== BEFORE CLEANING =====")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

print("\n===== AFTER CLEANING =====")
print("Rows:", messy_df.shape[0])
print("Columns:", messy_df.shape[1])
print("Missing Values:", messy_df.isnull().sum().sum())
print("Duplicate Rows:", messy_df.duplicated().sum())

print("\nFinal Columns:")
print(messy_df.columns.tolist())
```

```python
# Export cleaned dataset

messy_df.to_csv("clean_dataset.csv", index=False)

print("Clean dataset exported successfully!")
print("File name: clean_dataset.csv")
print("Final shape:", messy_df.shape)
```
